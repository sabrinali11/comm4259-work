### 1. Read data and obtain the shape

### 5. Rename column
Rename "'Revenue (in millions)" to be "Revenue_millions" and "Profit (in millions)" to be "Profit_millions"

```python
df.rename(columns = {
    "Revenue (in millions)": "Revenue_millions",
    "Profit (in millions)": "Profit_millions"
}, inplace = True)

df.head()
```

### 10. Get all rows with profit greater than 5,000 (millions) in 2000

### 15. Draw a histogram of profit in the year of 2000

```python
sns.histplot(
    df.loc[df.Year == 2000, "Profit_millions"].dropna(),
    kde = False
).set_title("Histogram of Profit in 2000")
```

### 16. Create a dataframe for Microsoft and Oracle. Draw a boxplot with the x axis showing two companies and y axis showing profit.

```python
df2 = df.loc[
    (df.Company == "Microsoft") |
    (df.Company == "Oracle")
]
```

```python
df2.head()
```

```python
sns.boxplot(
    x = "Company",
    y = "Profit_millions",
    data = df2
)
```

### 17. Draw two histograms side by side showing the profit for Microsoft and Oracle (FacetGrid)

```python
g = sns.FacetGrid(df2, col = "Company")
g = g.map(sns.histplot, "Profit_millions")
```

### 18. Draw a line plot for Microsoft and Oracle
Create a lineplot to show Microsoft and Oracle's profit trend by year. Check the lineplot usage about x, y, and hue [here](https://seaborn.pydata.org/generated/seaborn.lineplot.html)

```python
sns.lineplot(
    x = "Year",
    y = "Profit_millions",
    hue = "Company",
    data = df2
)
```
